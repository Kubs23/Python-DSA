```python
days = int(input("Enter the number of days: "))

seconds = days * 24 * 60 * 60

print("The number of seconds is:", seconds)
```

```python
import math

radius = float(input("Enter the radius: "))

volume = (4 / 3) * math.pi * (radius ** 3)

print("The volume of the sphere is:", volume)
```

```python
def area(side):
    return side ** 2


def perimeter(side):
    return 4 * side


side = float(input("Enter the side length of the square: "))

print("Area:", area(side))
print("Perimeter:", perimeter(side))
```

```python
def check_case(character):
    if character.isupper():
        print("The character is uppercase.")
    elif character.islower():
        print("The character is lowercase.")
    else:
        print("The character is not a letter.")


character = input("Enter a character: ")

check_case(character)
```

```python
x = 0
y = 20

while y >= 6:
    y -= 4
    x += 2 / y

print(x)
```

```python
values = []

for i in range(5):
    value = float(input("Enter a value: "))
    values.append(value)

average = sum(values) / len(values)

print("The values entered are:", values)
print("The average is:", average)
```
